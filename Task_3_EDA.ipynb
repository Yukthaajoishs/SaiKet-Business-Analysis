```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/Telco_Customer_Churn_Dataset  (3).csv")
```

```python
df.head()
```

```python
df.columns
```

```python
df.columns = df.columns.str.lower().str.replace(" ", "_")
```

```python
df.columns
```

```python
df["totalcharges"] = pd.to_numeric(df["totalcharges"], errors="coerce")
```

```python
df["totalcharges"].dtype
```

```python
df[["tenure", "monthlycharges", "totalcharges"]].mean()
```

```python
df[["tenure", "monthlycharges", "totalcharges"]].median()
```

```python
df[["tenure", "monthlycharges", "totalcharges"]].mode().iloc[0]
```

```python
summary = pd.DataFrame({
    "Mean": df[["tenure", "monthlycharges", "totalcharges"]].mean(),
    "Median": df[["tenure", "monthlycharges", "totalcharges"]].median(),
    "Mode": df[["tenure", "monthlycharges", "totalcharges"]].mode().iloc[0]
})

summary
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["tenure"], bins=20)

plt.xlabel("Tenure (Months)")
plt.ylabel("Number of Customers")
plt.title("Distribution of Customer Tenure")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["monthlycharges"], bins=20)

plt.xlabel("Monthly Charges")
plt.ylabel("Number of Customers")
plt.title("Distribution of Monthly Charges")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["totalcharges"], bins=20)

plt.xlabel("Total Charges")
plt.ylabel("Number of Customers")
plt.title("Distribution of Total Charges")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(x=df["tenure"])

plt.xlabel("Tenure (Months)")
plt.title("Box Plot of Customer Tenure")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(x=df["monthlycharges"])

plt.xlabel("Monthly Charges")
plt.title("Box Plot of Monthly Charges")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(x=df["totalcharges"])

plt.xlabel("Total Charges")
plt.title("Box Plot of Total Charges")

plt.show()
```
